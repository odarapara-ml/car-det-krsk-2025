## Creating a dataset for detecting cars in Krasnoyarsk 2025

The dataset is annotated for a single class: passenger cars. This decision was made because the project is intended as a learning exercise, and the dataset is relatively small (~500 images).

Focusing on a single, homogeneous class allows the model to learn more effectively and achieve higher detection accuracy. Including multiple vehicle types (such as buses, trucks, or motorcycles) in this small dataset could lead to poor generalization, as these objects differ significantly in shape and size, and there would be too few examples of each class for the model to learn reliably.

This approach ensures that the model can accurately detect passenger cars in various conditions, while keeping the project manageable and suitable for educational purposes. In future work, additional classes could be added once a larger, more diverse dataset is available.

image resolution : 1280x720

# Frame Extraction from Dashcam Recordings

To prepare the data, a Python script was implemented that:

- Sequentially loads all video files from the specified folder.
- Extracts frames at a fixed interval of 15 seconds.
- Rescales each frame to a unified resolution of 1280×720 pixels.
- Saves the frames into the images folder in .jpg format.
- Generates filenames in a unified structure.

After extraction, all images were manually reviewed, and irrelevant frames (e.g., black screens, nearly identical frames, or frames without any cars) were removed.

Example of filename structure:
video01_frame0001.jpg
video01_frame0002.jpg
video02_frame0001.jpg

```python
import cv2
import os

# Папка с видео
video_dir = r"C:\Users\odara\Downloads\car_det_krsk_dataset\videos\vk"
# Папка для сохранения кадров
output_dir = os.path.join(video_dir, "images")
os.makedirs(output_dir, exist_ok=True)

# Интервал в секундах
interval_sec = 15  

# Счётчик для названий видео
video_counter = 1

for filename in sorted(os.listdir(video_dir)):
    if not filename.lower().endswith((".mp4", ".avi", ".mov", ".mkv")):
        continue

    video_path = os.path.join(video_dir, filename)
    cap = cv2.VideoCapture(video_path)

    if not cap.isOpened():
        print(f"Не удалось открыть {filename}")
        continue

    fps = cap.get(cv2.CAP_PROP_FPS)
    if fps == 0:
        print(f"FPS = 0 у файла {filename}, пропускаем")
        continue

    frame_interval = int(fps * interval_sec)
    total_frames = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))

    base_name = f"video{video_counter:02d}"
    print(f" {filename}: {total_frames} кадров, FPS={fps}, интервал={frame_interval}")

    frame_num = 0
    saved = 0
    frame_counter = 1
    while True:
        ret, frame = cap.read()
        if not ret:
            break

        if frame_num % frame_interval == 0:
            out_name = f"{base_name}_frame{frame_counter:04d}.jpg"
            out_path = os.path.join(output_dir, out_name)

            # Приведём все кадры к одному размеру (1280x720)
            frame_resized = cv2.resize(frame, (1280, 720))
            cv2.imwrite(out_path, frame_resized)
            saved += 1
            frame_counter += 1

        frame_num += 1

    cap.release()
    print(f"{filename}: сохранено {saved} кадров")
    video_counter += 1  # увеличиваем номер видео

print("Извлечение кадров завершено!")
```
